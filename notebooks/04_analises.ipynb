```python
# ============================================================
# NOTEBOOK 04 - Analises de Negocio
# ============================================================

# Pergunta 1: Quais empresas aereas tiveram mais passageiros nos ultimos 10 anos?
spark.sql("""
SELECT
    EMPRESA_SIGLA,
    SUM(PASSAGEIROS_PAGOS) AS TOTAL_PASSAGEIROS,
    SUM(DECOLAGENS) AS TOTAL_VOOS
FROM workspace.default.gold_fato_voos
GROUP BY EMPRESA_SIGLA
ORDER BY TOTAL_PASSAGEIROS DESC
LIMIT 10
""").show()
```

```python
# Pergunta 2: Como evoluiu o trafego de passageiros ao longo dos anos?
spark.sql("""
SELECT
    ANO,
    SUM(PASSAGEIROS_PAGOS) AS TOTAL_PASSAGEIROS,
    SUM(DECOLAGENS) AS TOTAL_VOOS
FROM workspace.default.gold_fato_voos
GROUP BY ANO
ORDER BY ANO
""").show()
```

```python
# Pergunta 3: Quais as rotas mais movimentadas (por número de passageiros)?
spark.sql("""
SELECT
    ORIGEM,
    DESTINO,
    SUM(PASSAGEIROS_PAGOS) AS TOTAL_PASSAGEIROS,
    SUM(DECOLAGENS) AS TOTAL_VOOS
FROM workspace.default.gold_fato_voos
GROUP BY ORIGEM, DESTINO
ORDER BY TOTAL_PASSAGEIROS DESC
LIMIT 10
""").show()
```

```python
# Pergunta 4: Existe sazonalidade no trafego aereo? (padrao por mes)
spark.sql("""
SELECT
    MES,
    SUM(PASSAGEIROS_PAGOS) AS TOTAL_PASSAGEIROS,
    SUM(DECOLAGENS) AS TOTAL_VOOS,
    ROUND(SUM(PASSAGEIROS_PAGOS) / SUM(DECOLAGENS), 1) AS MEDIA_PASS_POR_VOO
FROM workspace.default.gold_fato_voos
GROUP BY MES
ORDER BY MES
""").show()
```

```python
# Pergunta 5: Quais empresas mais transportaram carga (kg) nos ultimos 10 anos?
spark.sql("""
SELECT
    EMPRESA_SIGLA,
    SUM(CARGA_PAGA_KG) AS TOTAL_CARGA_KG,
    ROUND(SUM(CARGA_PAGA_KG) / 1000000, 2) AS TOTAL_CARGA_TON,
    SUM(DECOLAGENS) AS TOTAL_VOOS
FROM workspace.default.gold_fato_voos
GROUP BY EMPRESA_SIGLA
ORDER BY TOTAL_CARGA_KG DESC
LIMIT 10
""").show()
```
