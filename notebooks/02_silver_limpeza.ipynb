```python
# ============================================================
# NOTEBOOK 02 - Silver Layer: Limpeza e Padronização
# ============================================================
from pyspark.sql import functions as F
from pyspark.sql.types import IntegerType

df_silver = spark.table("workspace.default.bronze_voos")
print("Bronze carregada:", df_silver.count(), "linhas")
print("Colunas:", df_silver.columns)
```

```python
spark.sql("""
CREATE OR REPLACE TABLE workspace.default.silver_voos AS
SELECT DISTINCT *
FROM workspace.default.bronze_voos
WHERE EMPRESA_SIGLA IS NOT NULL
AND AEROPORTO_DE_ORIGEM_SIGLA IS NOT NULL
AND AEROPORTO_DE_DESTINO_SIGLA IS NOT NULL
""")

print("Silver criada com sucesso!")
spark.sql("SELECT count(*) as total FROM workspace.default.silver_voos").show()
```
