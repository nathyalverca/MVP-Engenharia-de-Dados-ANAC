```python
# ============================================================
# NOTEBOOK 01 - Bronze Layer: Ingestão dos Dados Brutos
# ============================================================

arquivo_csv = "/Volumes/workspace/default/raw_data/Base_10_anos.csv"

df_bronze = (spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .option("sep", ";")
    .option("encoding", "UTF-8")
    .csv(arquivo_csv))

print("Linhas:", df_bronze.count())
print("Colunas:", len(df_bronze.columns))
df_bronze.printSchema()
```

```python
import re

def limpar_coluna(nome):
    nome = nome.upper().strip()
    nome = re.sub(r'[\s\(\)]+', '_', nome)
    nome = re.sub(r'[^A-Z0-9_]', '', nome)
    nome = re.sub(r'_+', '_', nome)
    return nome.strip('_')

df_bronze_clean = df_bronze
for col in df_bronze.columns:
    df_bronze_clean = df_bronze_clean.withColumnRenamed(col, limpar_coluna(col))

df_bronze_clean.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bronze_voos")
print("Tabela bronze_voos salva com sucesso!")
```
